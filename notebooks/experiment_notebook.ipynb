# Machine Unlearning Experiment

This notebook runs the end-to-end class-removal demo. The baseline model learns both classes, then the unlearning workflows remove all class-1 training examples and compare full retraining with the approximate retain-and-scrub method.

The important thing to watch is the removed-set accuracy: it starts high for the baseline model, then should drop after unlearning.

```python
# Let the notebook import the local machine_unlearning package whether it is
# launched from the project root or from inside the notebooks folder.
from pathlib import Path
import sys

from IPython.display import Image, display

project_root = Path.cwd()
if not (project_root / "machine_unlearning").exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

# run_experiment performs the full baseline -> retrain -> approximate unlearning pipeline.
from machine_unlearning.experiment import run_experiment
```

```python
# Run the demo and regenerate artifacts/results.json plus the visualization PNGs.
summary = run_experiment(project_root / "artifacts")

# Display the full metrics dictionary for inspection.
summary
```

```python
# Print the headline numbers that are easiest to explain in a presentation.
print(f"Removal strategy: {summary['dataset']['removal_strategy']}")
print(f"Removed points: {summary['dataset']['removed_points']}")
print(f"Full retraining removed accuracy drop: {summary['retrain_removed_accuracy_drop']:.1%}")
print(f"Approximate unlearning removed accuracy drop: {summary['approx_removed_accuracy_drop']:.1%}")
print(f"Approximate speedup vs retraining: {summary['speedup_retrain_over_approx']:.2f}x")
```
